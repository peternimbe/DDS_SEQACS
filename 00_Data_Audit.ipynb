# 00 Data Audit

Audits all five attached CSV files and confirms dataset integrity.

```python
from pathlib import Path
import sys, numpy as np, pandas as pd, matplotlib.pyplot as plt
ROOT=Path.cwd().parent if Path.cwd().name=="notebooks" else Path.cwd()
sys.path.insert(0,str(ROOT/"src"))
from dds_analysis import *
DATA=ROOT/"data"; OUT=ROOT/"outputs"; OUT.mkdir(exist_ok=True)
df=load_seqacs(DATA); validate_schema(df)
print(f"Loaded {len(df):,} rows and {len(df.columns)} columns")
```

```python
print(df.shape); display(df.head()); print(df.dtypes); print("\nMissing values:",int(df.isna().sum().sum()))
print("\nScenario counts:"); display(df["scenario"].value_counts().rename_axis("scenario").reset_index(name="n"))
print("\nDuplicate record IDs:",df["record_id"].duplicated().sum())
```

```python
audit=pd.DataFrame({"rows":[len(df)],"columns":[len(df.columns)],"missing_cells":[df.isna().sum().sum()],
"duplicate_record_ids":[df["record_id"].duplicated().sum()]})
audit.to_csv(OUT/"data_audit.csv",index=False)
```
